# Bài 3 (Trang 7): Neural Network trên MNIST (k-fold cross validation)
**Họ tên:** Trần Minh Tấn - **MSSV:** 6451071071

## Chuẩn bị dữ liệu (câu a, b, c - cần để chạy câu d, e)

```python
from tensorflow.keras.datasets import mnist

# a) Đọc dữ liệu
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

# c) Chuẩn hóa pixel về đoạn [0, 1]
train_images = train_images.astype("float32") / 255.0
test_images = test_images.astype("float32") / 255.0

print("Họ tên: Trần Minh Tấn - MSSV: 6451071071")
print("train_images:", train_images.shape, "| train_labels:", train_labels.shape)
print("test_images:", test_images.shape, "| test_labels:", test_labels.shape)
print("Giá trị pixel nhỏ nhất/lớn nhất:", train_images.min(), "/", train_images.max())
```

## Câu d) Định nghĩa hàm create_model

```python
from tensorflow.keras import models, layers

def create_model():
    model = models.Sequential([
        layers.Input(shape=(28, 28)),
        layers.Flatten(),
        layers.Dense(128, activation="relu"),
        layers.Dense(10, activation="softmax")
    ])
    model.compile(optimizer="adam",
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])
    return model

print("Họ tên: Trần Minh Tấn - MSSV: 6451071071")
create_model().summary()
```

## Câu e) Huấn luyện mô hình bằng k-fold cross validation (k = 10)

```python
import numpy as np
from sklearn.model_selection import KFold

print("Họ tên: Trần Minh Tấn - MSSV: 6451071071")

kfold = KFold(n_splits=10, shuffle=True, random_state=42)

models_list = []         # lưu model sau mỗi fold
accuracy_per_fold = []
loss_per_fold = []

fold_no = 1
for train_index, val_index in kfold.split(train_images):
    print(f"\n===== Fold {fold_no}/10 =====")

    model = create_model()

    X_train_fold, X_val_fold = train_images[train_index], train_images[val_index]
    y_train_fold, y_val_fold = train_labels[train_index], train_labels[val_index]

    model.fit(X_train_fold, y_train_fold,
              epochs=5, batch_size=32,
              validation_data=(X_val_fold, y_val_fold),
              verbose=1)

    models_list.append(model)

    loss, accuracy = model.evaluate(X_val_fold, y_val_fold, verbose=0)
    accuracy_per_fold.append(accuracy)
    loss_per_fold.append(loss)
    print(f"Fold {fold_no} - Loss: {loss:.4f} - Accuracy: {accuracy:.4f}")

    fold_no += 1

print("\nHọ tên: Trần Minh Tấn - MSSV: 6451071071")
print("Accuracy từng fold:", [round(a, 4) for a in accuracy_per_fold])
print("Loss từng fold:", [round(l, 4) for l in loss_per_fold])
print(f"Accuracy trung bình: {np.mean(accuracy_per_fold):.4f}")
print(f"Loss trung bình: {np.mean(loss_per_fold):.4f}")
```
