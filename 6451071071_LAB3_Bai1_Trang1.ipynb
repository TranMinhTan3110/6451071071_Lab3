Cell 1: Câu a (đọc dữ liệu)

```python
from sklearn.datasets import load_iris

iris = load_iris()

print("Họ tên: Trần Minh Tấn - MSSV: 6451071071")
print("Câu a: Đã đọc bộ dữ liệu Iris từ sklearn")
```

Cell 2: Câu b, in tên thuộc tính và thuộc tính phân lớp

```python
print("Họ tên: Trần Minh Tấn - MSSV: 6451071071")
print("Tên các thuộc tính:", iris.feature_names)
print("Tên các lớp (thuộc tính phân lớp):", iris.target_names)
```

Cell 3: Số lượng mẫu, số lượng thuộc tính

```python
print("Họ tên: Trần Minh Tấn - MSSV: 6451071071")
print("data.shape:", iris.data.shape)
print("target.shape:", iris.target.shape)
print("Số lượng mẫu:", iris.data.shape[0])
print("Số lượng thuộc tính:", iris.data.shape[1])
```

Cell 4: In 5 mẫu dữ liệu đầu tiên

```python


print("Họ tên: Trần Minh Tấn - MSSV: 6451071071")
print("5 mẫu data đầu tiên:")
print(iris.data[:5])
print("5 nhãn target đầu tiên:")
print(iris.target[:5])
```

Cell 5: Thống kê count, mean, std, min, max, tứ phân vị

```python
import pandas as pd

print("Họ tên: Trần Minh Tấn - MSSV: 6451071071")
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target
print(df.describe())
```
