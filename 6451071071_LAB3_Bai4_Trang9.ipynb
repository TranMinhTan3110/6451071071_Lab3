# Bài 4 (Trang 9): Neural Network trên bộ dữ liệu Iris (k-fold cross validation)
**Họ tên:** Trần Minh Tấn - **MSSV:** 6451071071

## Chuẩn bị dữ liệu (câu a, b - cần để chạy câu c, d)

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler

# a) Đọc bộ dữ liệu Iris
iris = load_iris()
X = iris.data
y = iris.target

# b) Chuẩn hóa data về đoạn [0, 1]
scaler = MinMaxScaler()
X = scaler.fit_transform(X)

print("Họ tên: Trần Minh Tấn - MSSV: 6451071071")
print("X:", X.shape, "| y:", y.shape)
print("Giá trị nhỏ nhất/lớn nhất sau chuẩn hóa:", X.min(), "/", X.max())
```

## Câu c) Định nghĩa hàm create_model

```python
from tensorflow.keras import models, layers

def create_model():
    model = models.Sequential([
        layers.Input(shape=(4,)),
        layers.Dense(16, activation="relu"),
        layers.Dense(8, activation="relu"),
        layers.Dense(3, activation="softmax")
    ])
    model.compile(optimizer="adam",
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])
    return model

print("Họ tên: Trần Minh Tấn - MSSV: 6451071071")
create_model().summary()
```

## Câu d) Huấn luyện mô hình bằng k-fold cross validation (k = 10)

```python
import numpy as np
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print("Họ tên: Trần Minh Tấn - MSSV: 6451071071")

kfold = KFold(n_splits=10, shuffle=True, random_state=42)

models_list = []          # lưu model sau mỗi fold
accuracy_per_fold = []
precision_per_fold = []
recall_per_fold = []
f1_per_fold = []

fold_no = 1
for train_index, val_index in kfold.split(X):
    X_train_fold, X_val_fold = X[train_index], X[val_index]
    y_train_fold, y_val_fold = y[train_index], y[val_index]

    model = create_model()
    model.fit(X_train_fold, y_train_fold, epochs=100, batch_size=16, verbose=0)
    models_list.append(model)

    y_pred = np.argmax(model.predict(X_val_fold, verbose=0), axis=1)

    acc = accuracy_score(y_val_fold, y_pred)
    pre = precision_score(y_val_fold, y_pred, average="macro", zero_division=0)
    rec = recall_score(y_val_fold, y_pred, average="macro", zero_division=0)
    f1 = f1_score(y_val_fold, y_pred, average="macro", zero_division=0)

    accuracy_per_fold.append(acc)
    precision_per_fold.append(pre)
    recall_per_fold.append(rec)
    f1_per_fold.append(f1)

    print(f"Fold {fold_no:2d} - Accuracy: {acc:.4f} - Precision: {pre:.4f} - Recall: {rec:.4f} - F1-score: {f1:.4f}")
    fold_no += 1

print("\nHọ tên: Trần Minh Tấn - MSSV: 6451071071")
print(f"Accuracy trung bình : {np.mean(accuracy_per_fold):.4f}")
print(f"Precision trung bình: {np.mean(precision_per_fold):.4f}")
print(f"Recall trung bình   : {np.mean(recall_per_fold):.4f}")
print(f"F1-score trung bình : {np.mean(f1_per_fold):.4f}")
```
